Q2 Collatz

```python
col=[] #final step list
values=[]
for i in range(2,100):
    steps=[] #temporary step list
    k=i #number is defined to manipulate
    while k!=1: #stop when done
        if k%2==0: #collatz
            k=k/2
            steps.append(k) #step counter, which also tells value of that step
        else:
            k=3*k+1
            steps.append(k)
    col.append(len(steps)) #the length of steps must be total steps
    values=values+steps #lines up all step-values from a certain starting number
    

print("The most steps:",max(col))
print("occur from number:",col.index(max(col))+1) #zero indexing

print("higest value assumed is:",max(values))
```
